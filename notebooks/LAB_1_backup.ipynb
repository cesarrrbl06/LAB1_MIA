## 2. Noise Addition  (Álvaro Gil Burgo)
### 2.1 Gaussian Noise


```python
def show_gaussian(img, name, seed, filename):
    rng = np.random.default_rng(seed)
    versions = {"Original": img}
    for level, sigma in GAUSS_LEVELS.items():
        versions[f"sigma={sigma} ({level})"] = add_gaussian_noise(img, sigma, rng)

    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    for col, (title, im) in enumerate(versions.items()):
        axes[0, col].imshow(im, cmap="gray", vmin=0, vmax=1)
        p = "" if title == "Original" else f"\nPSNR={psnr(img, im, data_range=1.0):.1f} dB"
        axes[0, col].set_title(title + p)
        axes[0, col].axis("off")

        axes[1, col].hist(im.ravel(), bins=256, range=(0, 1), color="gray", log=True)
        axes[1, col].set_xlabel("Intensity")
        axes[1, col].set_ylabel("Pixel count (log)")
    fig.suptitle(f"Gaussian Noise: {name}")
    plt.tight_layout()
    save_figure(fig, filename)
    plt.show()

show_gaussian(img1, "Brain MRI", seed=100, filename="01_gaussian_noise_mri.png")
show_gaussian(img2, "Chest X-ray", seed=101, filename="01_gaussian_noise_xray.png")
```

## 3. Standard Smoothing Methods (Álvaro Gil Burgo)
Evaluation of Mean, Gaussian, and Median filters on corrupted images.


```python
from pathlib import Path

TABLE_DIR = Path("results/tables")
TABLE_DIR.mkdir(parents=True, exist_ok=True)
RECOMPUTE = False      # ponlo a True para forzar que se recalculen las búsquedas


CLEAN = {"Brain MRI": img1, "Chest X-ray": img2}
SEEDS = {"gaussian": (100, 101), "impulse": (200, 201)}   # (image 1, image 2)
NOISE_CFG = {
    "gaussian": (GAUSS_LEVELS, add_gaussian_noise),
    "impulse": (IMPULSE_LEVELS, add_impulse_noise),
}

def build_noisy_dataset():
    """Returns {(image, noise_type, level): noisy_image} -> 2 x 2 x 3 = 12 images."""
    data = {}
    for noise_type, (levels, add_fn) in NOISE_CFG.items():
        for (name, clean), seed in zip(CLEAN.items(), SEEDS[noise_type]):
            rng = np.random.default_rng(seed)
            for level, value in levels.items():
                data[(name, noise_type, level)] = add_fn(clean, value, rng=rng)
    return data

NOISY = build_noisy_dataset()
```

```python
FILTERS = {"mean": filter_mean, "gaussian": filter_gaussian, "median": filter_median}
PARAM_NAME = {"mean": "size", "gaussian": "sigma", "median": "size"}
PARAM_GRID = {
    "mean": [3, 5, 7],
    "gaussian": [0.3, 0.5, 0.75, 1.0, 1.5, 2.0, 2.5, 3.0],
    "median": [3, 5, 7],
}
```

```python
from skimage import transform

def downscale(img, max_side=512):
    """Reduce la imagen para que su lado mayor mida max_side, conservando la proporcion."""
    if max(img.shape) <= max_side:
        return img
    return np.clip(transform.rescale(img, max_side / max(img.shape), anti_aliasing=True), 0, 1)

img1 = downscale(img1)
img2 = downscale(img2)
print("Dimensiones reescaladas:", img1.shape, img2.shape)

# Actualizar CLEAN y NOISY con las versiones reescaladas (evita procesar 2412x1956 pixeles)
CLEAN = {"Brain MRI": img1, "Chest X-ray": img2}
NOISY = build_noisy_dataset()
```

```python
SWEEP_CSV = TABLE_DIR / "03_standard_filters_sweep.csv"

if SWEEP_CSV.exists() and not RECOMPUTE:
    sweep = pd.read_csv(SWEEP_CSV)
else:
    rows = []
    for (name, noise_type, level), noisy in NOISY.items():
        for fname, fn in FILTERS.items():
            for p in PARAM_GRID[fname]:
                m = evaluate_quality(CLEAN[name], fn(noisy, p))
                rows.append({"image": name, "noise": noise_type, "level": level,
                             "filter": fname, "param": p,
                             "PSNR": m["PSNR (dB)"], "SSIM": m["SSIM"]})
    sweep = pd.DataFrame(rows)
    sweep.to_csv(SWEEP_CSV, index=False)
```

```python
KEYS = ["image", "noise", "level", "filter"]
LEVEL_ORDER = {"low": 0, "medium": 1, "high": 2}

best = sweep.loc[sweep.groupby(KEYS)["PSNR"].idxmax()].rename(columns={"param": "best_param"})
best["_o"] = best["level"].map(LEVEL_ORDER)
best = best.sort_values(["image", "noise", "_o", "filter"]).drop(columns="_o").reset_index(drop=True)
# Optimum on the border of the grid -> the grid may need to be extended
best["at_grid_edge"] = [
    r.best_param == PARAM_GRID[r["filter"]][-1]
    or (r["filter"] == "gaussian" and r.best_param == PARAM_GRID[r["filter"]][0])
    for _, r in best.iterrows()
]
display(best)
```

```python
def plot_psnr_sweeps(image="Brain MRI"):
    noises = list(NOISE_CFG)
    fig, axes = plt.subplots(len(noises), 3, figsize=(15, 3.6 * len(noises)), squeeze=False)
    for r, noise_type in enumerate(noises):
        for c, fname in enumerate(FILTERS):
            ax = axes[r, c]
            for level in LEVEL_ORDER:
                d = sweep[(sweep.image == image) & (sweep.noise == noise_type)
                          & (sweep["filter"] == fname) & (sweep.level == level)]
                ax.plot(d["param"], d["PSNR"], marker="o", label=level)
                k = d["PSNR"].idxmax()
                ax.plot(d.loc[k, "param"], d.loc[k, "PSNR"], marker="*", color="k", ms=11)
            ax.set_title(f"{image} | {noise_type} | {fname}")
            ax.set_xlabel(PARAM_NAME[fname]); ax.set_ylabel("PSNR (dB)"); ax.grid(alpha=0.3)
            if r == 0 and c == 0:
                ax.legend(title="noise level")
    plt.tight_layout()
    save_figure(fig, f"03_standard_sweeps_psnr_{image.split()[0].lower()}.png")
    plt.show()

plot_psnr_sweeps("Brain MRI")
```

```python
def best_param(name, noise_type, level, fname):
    sel = best[(best.image == name) & (best.noise == noise_type)
               & (best.level == level) & (best["filter"] == fname)]
    p = sel["best_param"].iloc[0]
    return int(p) if PARAM_NAME[fname] == "size" else float(p)   # window sizes must be int

def show_example(name="Brain MRI", level="medium"):
    """One figure: clean, noisy and the 3 filters (best parameters) for both noise types."""
    fig, axes = plt.subplots(len(NOISE_CFG), 5, figsize=(20, 4.4 * len(NOISE_CFG)))
    for r, noise_type in enumerate(NOISE_CFG):
        noisy = NOISY[(name, noise_type, level)]
        panels = [("Clean", CLEAN[name]), (f"{noise_type} noise ({level})", noisy)]
        for fname, fn in FILTERS.items():
            p = best_param(name, noise_type, level, fname)
            m = evaluate_quality(CLEAN[name], fn(noisy, p))
            panels.append((f"{fname} ({PARAM_NAME[fname]}={p})\nPSNR={m['PSNR (dB)']:.2f} dB  SSIM={m['SSIM']:.3f}",
                           fn(noisy, p)))
        for c, (title, im) in enumerate(panels):
            axes[r, c].imshow(im, cmap="gray", vmin=0, vmax=1)
            axes[r, c].set_title(title, fontsize=10); axes[r, c].axis("off")
    fig.suptitle(f"Standard smoothing, best parameters: {name}, {level} noise", fontsize=14)
    plt.tight_layout()
    save_figure(fig, f"03_standard_example_{name.split()[0].lower()}_{level}.png")
    plt.show()

show_example("Brain MRI", "medium")
```

## 4. Non-Local Means (NLM) Filtering (Álvaro Gil Burgo)
Non-local patch-based weighted averaging.


```python
H_GRID = [0.005, 0.02, 0.04, 0.08, 0.12, 0.24]   # absolute h
PATCH_SIZES = [3, 5]
PATCH_DISTANCES = [3, 6]
START_PATCH = (5, 6)           # (patch_size, patch_distance) used in stage 1
```

```python
nlm_rows = []

def run_nlm(name, noise_type, level, noisy, ps, pdist, h, stage):
    m = evaluate_quality(CLEAN[name], filter_nlm(noisy, patch_size=ps, patch_distance=pdist, h=h))
    nlm_rows.append({"image": name, "noise": noise_type, "level": level, "stage": stage,
                     "h": h, "patch_size": ps, "patch_distance": pdist,
                     "PSNR": m["PSNR (dB)"], "SSIM": m["SSIM"]})
    return m["PSNR (dB)"]

for (name, noise_type, level), noisy in NOISY.items():
    # Stage 1: sweep h with the patch fixed
    ps, pdist = START_PATCH
    s1 = {h: run_nlm(name, noise_type, level, noisy, ps, pdist, h, "1_h") for h in H_GRID}
    h_best = max(s1, key=s1.get)

    # Stage 2: with the best h, sweep patch_size x patch_distance
    s2 = {(a, b): run_nlm(name, noise_type, level, noisy, a, b, h_best, "2_patch")
          for a in PATCH_SIZES for b in PATCH_DISTANCES}
    ps, pdist = max(s2, key=s2.get)

    # Stage 3: with the best patch, re-check h in the neighbouring grid values
    i = H_GRID.index(h_best)
    for h in H_GRID[max(0, i - 1): i + 2]:
        run_nlm(name, noise_type, level, noisy, ps, pdist, h, "3_refine_h")
    print("done:", name, noise_type, level)

nlm_sweep = pd.DataFrame(nlm_rows)
```

```python
def wide(metric):
    w = best.pivot_table(index=SCEN, columns="filter", values=metric)
    w["NLM"] = nlm_best.set_index(SCEN)[metric]
    w = w.reindex(pd.MultiIndex.from_tuples(list(NOISY.keys()), names=SCEN))
    w["best_method"] = w[["mean", "gaussian", "median", "NLM"]].idxmax(axis=1)
    w["NLM_minus_best_std"] = w["NLM"] - w[["mean", "gaussian", "median"]].max(axis=1)
    return w.round(3)

print("PSNR (dB), best parameters of each method:")
display(wide("PSNR"))
print("SSIM, at the parameters chosen by PSNR:")
display(wide("SSIM"))
```

```python
def show_nlm_example(name="Brain MRI", level="medium"):
    """Clean, noisy, best standard filter and NLM (best parameters), for both noise types."""
    wide_psnr = wide("PSNR")
    fig, axes = plt.subplots(len(NOISE_CFG), 4, figsize=(16, 4.4 * len(NOISE_CFG)))
    for r, noise_type in enumerate(NOISE_CFG):
        noisy = NOISY[(name, noise_type, level)]
        std_name = wide_psnr.loc[(name, noise_type, level), ["mean", "gaussian", "median"]].astype(float).idxmax()
        p = best_param(name, noise_type, level, std_name)
        std_img = FILTERS[std_name](noisy, p)
        nb = nlm_best[(nlm_best.image == name) & (nlm_best.noise == noise_type) & (nlm_best.level == level)].iloc[0]
        nlm_img = filter_nlm(noisy, patch_size=int(nb.patch_size), patch_distance=int(nb.patch_distance), h=float(nb.h))
        m_std, m_nlm = evaluate_quality(CLEAN[name], std_img), evaluate_quality(CLEAN[name], nlm_img)
        panels = [
            ("Clean", CLEAN[name]),
            (f"{noise_type} noise ({level})", noisy),
            (f"{std_name} ({PARAM_NAME[std_name]}={p})\nPSNR={m_std['PSNR (dB)']:.2f} dB  SSIM={m_std['SSIM']:.3f}", std_img),
            (f"NLM (h={nb.h}, patch={int(nb.patch_size)}, dist={int(nb.patch_distance)})\n"
             f"PSNR={m_nlm['PSNR (dB)']:.2f} dB  SSIM={m_nlm['SSIM']:.3f}", nlm_img),
        ]
        for c, (title, im) in enumerate(panels):
            axes[r, c].imshow(im, cmap="gray", vmin=0, vmax=1)
            axes[r, c].set_title(title, fontsize=10); axes[r, c].axis("off")
    fig.suptitle(f"NLM vs best standard filter: {name}, {level} noise", fontsize=14)
    plt.tight_layout()
    save_figure(fig, f"04_nlm_example_{name.split()[0].lower()}_{level}.png")
    plt.show()

show_nlm_example("Brain MRI", "medium")
```

## 5. Anisotropic Diffusion (Perona and Malik) (César Manuel Robalo Fariñas)
Gradient-adaptive diffusion for edge preservation.


```python

PM_SWEEP_CSV = TABLE_DIR / "05_perona_malik_sweep.csv"
KAPPA_GRID = [0.03, 0.08, 0.15]
N_ITER_GRID = [10, 20]
OPTION_GRID = [1, 2]
GAMMA = 0.20

if PM_SWEEP_CSV.exists() and not RECOMPUTE:
    print(f"Loading cached results from: {PM_SWEEP_CSV}")
    pm_sweep = pd.read_csv(PM_SWEEP_CSV)
else:
    print("Executing Perona-Malik parameter sweep...")
    rows = []
    total_scenarios = len(NOISY)
    scen_idx = 0
    
    # Iterate across all noise degradation scenarios
    for (name, noise_type, level), noisy in NOISY.items():
        scen_idx += 1
        print(f"  [{scen_idx}/{total_scenarios}] {name} | {noise_type} ({level})...")
        
        # Grid search over conductivity option, kappa threshold, and iteration count
        for opt in OPTION_GRID:
            for kappa in KAPPA_GRID:
                for n_iter in N_ITER_GRID:
                    # Applies gradient-adaptive anisotropic diffusion filtering
                    filtered = filter_perona_malik(
                        noisy,
                        n_iter=n_iter,
                        kappa=kappa,
                        gamma=GAMMA,
                        option=opt
                    )
                    m = evaluate_quality(CLEAN[name], filtered)
                    rows.append({
                        "image": name,
                        "noise": noise_type,
                        "level": level,
                        "option": opt,
                        "kappa": kappa,
                        "n_iter": n_iter,
                        "PSNR": m["PSNR (dB)"],
                        "SSIM": m["SSIM"]
                    })
                    
    pm_sweep = pd.DataFrame(rows)
    pm_sweep.to_csv(PM_SWEEP_CSV, index=False)
    print(f"Parameter sweep saved to: {PM_SWEEP_CSV}")

# Extract optimal parameter combination maximizing PSNR per scenario
SCEN = ["image", "noise", "level"]
pm_best = pm_sweep.loc[pm_sweep.groupby(SCEN)["PSNR"].idxmax()].reset_index(drop=True)

print("\n--- Optimal Perona-Malik Parameters per Clinical Scenario ---")
display(pm_best[["image", "noise", "level", "option", "kappa", "n_iter", "PSNR", "SSIM"]])

# Build quantitative PSNR benchmark table across methods
comp_psnr = best.pivot_table(index=SCEN, columns="filter", values="PSNR")
if "nlm_best" in globals():
    comp_psnr["NLM"] = nlm_best.set_index(SCEN)["PSNR"]
comp_psnr["Perona-Malik"] = pm_best.set_index(SCEN)["PSNR"]
comp_psnr = comp_psnr.reindex(pd.MultiIndex.from_tuples(list(NOISY.keys()), names=SCEN))
comp_psnr["best_method"] = comp_psnr.idxmax(axis=1)

print("\n--- Quantitative PSNR (dB) Comparison Across Filtering Methods ---")
display(comp_psnr.round(2))


def show_pm_example(name="Brain MRI", level="medium"):
    """Comparative visual result: Clean, Noisy, Best Standard Filter, and Perona-Malik."""
    fig, axes = plt.subplots(len(NOISE_CFG), 4, figsize=(16, 4.4 * len(NOISE_CFG)))
    
    for r, noise_type in enumerate(NOISE_CFG):
        noisy = NOISY[(name, noise_type, level)]
        
        # 1. Retrieve best baseline standard filter
        std_sel = best[(best.image == name) & (best.noise == noise_type) & (best.level == level)]
        std_row = std_sel.loc[std_sel["PSNR"].idxmax()]
        std_name = std_row["filter"]
        p = std_row["best_param"]
        p_val = int(p) if std_name in ["mean", "median"] else float(p)
        
        # Applies standard spatial filter (Mean, Median, or Gaussian)
        std_img = FILTERS[std_name](noisy, p_val)
        m_std = evaluate_quality(CLEAN[name], std_img)
        
        # 2. Retrieve best Perona-Malik filter configuration
        pm_row = pm_best[(pm_best.image == name) & (pm_best.noise == noise_type) & (pm_best.level == level)].iloc[0]
        
        # Applies optimal Perona-Malik anisotropic diffusion filtering
        pm_img = filter_perona_malik(
            noisy,
            n_iter=int(pm_row.n_iter),
            kappa=float(pm_row.kappa),
            gamma=GAMMA,
            option=int(pm_row.option)
        )
        m_pm = evaluate_quality(CLEAN[name], pm_img)
        
        opt_lbl = "Exp" if pm_row.option == 1 else "Rat"
        panels = [
            ("Original (Clean)", CLEAN[name]),
            (f"{noise_type.capitalize()} noise ({level})", noisy),
            (f"Standard: {std_name} ({PARAM_NAME[std_name]}={p})\nPSNR={m_std['PSNR (dB)']:.2f} dB | SSIM={m_std['SSIM']:.3f}", std_img),
            (f"Perona-Malik (opt={opt_lbl}, k={pm_row.kappa}, it={int(pm_row.n_iter)})\nPSNR={m_pm['PSNR (dB)']:.2f} dB | SSIM={m_pm['SSIM']:.3f}", pm_img)
        ]
        
        for c, (title, im) in enumerate(panels):
            axes[r, c].imshow(im, cmap="gray", vmin=0, vmax=1)
            axes[r, c].set_title(title, fontsize=10)
            axes[r, c].axis("off")
            
    fig.suptitle(f"Perona-Malik vs Best Standard Filter: {name} ({level} noise)", fontsize=14)
    plt.tight_layout()
    save_figure(fig, f"05_perona_malik_example_{name.split()[0].lower()}_{level}.png")
    plt.show()

show_pm_example("Brain MRI", "medium")
```

### Comment on the code 

Unlike isotropic filters that apply uniform blurring and degrade spatial resolution at tissue boundaries, Perona-Malik anisotropic diffusion adaptively controls the smoothing process through a gradient-dependent conductivity function $c(\Vert{}\nabla I\Vert{})$, which suppresses diffusion across strong anatomical edges while removing noise within homogeneous regions[cite: 2]. In this section, we perform a systematic grid search over key hyperparameters—the edge sensitivity threshold ($\kappa \in \{0.03, 0.08, 0.15\}$), iteration depth ($N_{\text{iter}} \in \{10, 20\}$), and conductivity formulation (`option=1` for exponential decay, favoring homogeneous continuity, vs. `option=2` for rational decay, preserving sharper structural transitions) while maintaining the integration step at $\gamma = 0.20$ to ensure numerical stability on discrete 2D grids

Exploration and benchmarking of a state-of-the-art literature method.


```python
ADV_SWEEP_CSV = TABLE_DIR / "06_advanced_bilateral_sweep.csv"
SIGMA_COLOR_GRID = [0.03, 0.08, 0.15]
SIGMA_SPATIAL_GRID = [1.0, 2.0]
WIN_SIZE = 5


if ADV_SWEEP_CSV.exists() and not RECOMPUTE:
    print(f"Loading cached results from: {ADV_SWEEP_CSV}")
    adv_sweep = pd.read_csv(ADV_SWEEP_CSV)
else:
    print("Executing Bilateral Filter parameter sweep...")
    rows = []
    total_scenarios = len(NOISY)
    scen_idx = 0
    
    # Iterate across all noise degradation scenarios
    for (name, noise_type, level), noisy in NOISY.items():
        scen_idx += 1
        print(f"  [{scen_idx}/{total_scenarios}] {name} | {noise_type} ({level})...")
        
        # Grid search over range sigma (color) and spatial sigma combinations
        for sc in SIGMA_COLOR_GRID:
            for ss in SIGMA_SPATIAL_GRID:
                # Applies bilateral filtering (combines spatial and range Gaussian kernels)
                filtered = filter_bilateral(
                    noisy,
                    win_size=WIN_SIZE,
                    sigma_color=sc,
                    sigma_spatial=ss
                )
                m = evaluate_quality(CLEAN[name], filtered)
                rows.append({
                    "image": name,
                    "noise": noise_type,
                    "level": level,
                    "win_size": WIN_SIZE,
                    "sigma_color": sc,
                    "sigma_spatial": ss,
                    "PSNR": m["PSNR (dB)"],
                    "SSIM": m["SSIM"]
                })
                
    adv_sweep = pd.DataFrame(rows)
    adv_sweep.to_csv(ADV_SWEEP_CSV, index=False)
    print(f"Parameter sweep saved to: {ADV_SWEEP_CSV}")


# Extract optimal bilateral parameter combination maximizing PSNR per scenario
SCEN = ["image", "noise", "level"]
adv_best = adv_sweep.loc[adv_sweep.groupby(SCEN)["PSNR"].idxmax()].reset_index(drop=True)

print("\n--- Optimal Bilateral Filter Parameters per Clinical Scenario ---")
display(adv_best[["image", "noise", "level", "sigma_color", "sigma_spatial", "PSNR", "SSIM"]])

# Build comprehensive PSNR benchmark table across all investigated methods
comp_all = best.pivot_table(index=SCEN, columns="filter", values="PSNR")
if "nlm_best" in globals():
    comp_all["NLM"] = nlm_best.set_index(SCEN)["PSNR"]
if "pm_best" in globals():
    comp_all["Perona-Malik"] = pm_best.set_index(SCEN)["PSNR"]
comp_all["Bilateral"] = adv_best.set_index(SCEN)["PSNR"]
comp_all = comp_all.reindex(pd.MultiIndex.from_tuples(list(NOISY.keys()), names=SCEN))
comp_all["best_method"] = comp_all.idxmax(axis=1)

print("\n--- Global Quantitative PSNR (dB) Comparison Across All Methods ---")
display(comp_all.round(2))


def show_adv_example(name="Brain MRI", level="medium"):
    """Comparative visual result: Clean, Noisy, Best Standard Filter, and Bilateral Filter."""
    fig, axes = plt.subplots(len(NOISE_CFG), 4, figsize=(16, 4.4 * len(NOISE_CFG)))
    
    for r, noise_type in enumerate(NOISE_CFG):
        noisy = NOISY[(name, noise_type, level)]
        
        # 1. Retrieve best baseline standard filter
        std_sel = best[(best.image == name) & (best.noise == noise_type) & (best.level == level)]
        std_row = std_sel.loc[std_sel["PSNR"].idxmax()]
        std_name = std_row["filter"]
        p = std_row["best_param"]
        p_val = int(p) if std_name in ["mean", "median"] else float(p)
        
        # Applies baseline spatial filter (Mean, Median, or Gaussian)
        std_img = FILTERS[std_name](noisy, p_val)
        m_std = evaluate_quality(CLEAN[name], std_img)
        
        # 2. Retrieve best Bilateral Filter configuration
        b_row = adv_best[(adv_best.image == name) & (adv_best.noise == noise_type) & (adv_best.level == level)].iloc[0]
        
        # Applies optimal Bilateral filtering
        bilat_img = filter_bilateral(
            noisy,
            win_size=int(b_row.win_size),
            sigma_color=float(b_row.sigma_color),
            sigma_spatial=float(b_row.sigma_spatial)
        )
        m_b = evaluate_quality(CLEAN[name], bilat_img)
        
        panels = [
            ("Original (Clean)", CLEAN[name]),
            (f"{noise_type.capitalize()} noise ({level})", noisy),
            (f"Standard: {std_name} ({PARAM_NAME[std_name]}={p})\nPSNR={m_std['PSNR (dB)']:.2f} dB | SSIM={m_std['SSIM']:.3f}", std_img),
            (f"Bilateral (sc={b_row.sigma_color}, ss={b_row.sigma_spatial})\nPSNR={m_b['PSNR (dB)']:.2f} dB | SSIM={m_b['SSIM']:.3f}", bilat_img)
        ]
        
        for c, (title, im) in enumerate(panels):
            axes[r, c].imshow(im, cmap="gray", vmin=0, vmax=1)
            axes[r, c].set_title(title, fontsize=10)
            axes[r, c].axis("off")
            
    fig.suptitle(f"Bilateral Filter vs Best Standard Filter: {name} ({level} noise)", fontsize=14)
    plt.tight_layout()
    save_figure(fig, f"06_bilateral_example_{name.split()[0].lower()}_{level}.png")
    plt.show()

show_adv_example("Brain MRI", "medium")
```

#### Section 6: Code Comment
The bilateral filter extends standard Gaussian spatial smoothing by combining a spatial Gaussian kernel ($\sigma_{\text{spatial}}$), which penalizes pixel distance, with an intensity range Gaussian kernel ($\sigma_{\text{color}}$), which penalizes radiometric pixel differences. This dual-weighting mechanism allows the filter to smooth out noise within uniform tissue regions while strictly maintaining sharp intensity boundaries. In this section, we perform a hyperparameter grid search over the range parameter ($\sigma_{\text{color}} \in \{0.03, 0.08, 0.15\}$) and spatial spread ($\sigma_{\text{spatial}} \in \{1.0, 2.0\}$) across a fixed window size ($W = 5$) to identify the optimal configuration that maximizes PSNR across all clinical noise scenarios.

## 7. Comparative Analysis and Metrics Summary (César Manuel Robalo Fariñas)
Consolidated benchmarking of PSNR and SSIM across all filtering methods.


```python

SCEN = ["image", "noise", "level"]
LEVEL_ORDER = ["low", "medium", "high"]


def get_best_df(var_name, csv_name, param_cols):
    """Retrieves optimal performance DataFrame from global scope or cached CSV."""
    if var_name in globals():
        return globals()[var_name]
    p = TABLE_DIR / csv_name
    if p.exists():
        df = pd.read_csv(p)
        return df.loc[df.groupby(SCEN)["PSNR"].idxmax()].reset_index(drop=True)
    return None


# Retrieve optimal scenarios per filtering family
b_std = best if "best" in globals() else None
b_nlm = get_best_df("nlm_best", "04_nlm_sweep.csv", ["h", "patch_size", "patch_distance"])
b_pm = get_best_df("pm_best", "05_perona_malik_sweep.csv", ["option", "kappa", "n_iter"])
b_adv = get_best_df("adv_best", "06_advanced_bilateral_sweep.csv", ["sigma_color", "sigma_spatial"])


# Construct consolidated PSNR master table
master_psnr = b_std.pivot_table(index=SCEN, columns="filter", values="PSNR")
if b_nlm is not None:
    master_psnr["NLM"] = b_nlm.set_index(SCEN)["PSNR"]
if b_pm is not None:
    master_psnr["Perona-Malik"] = b_pm.set_index(SCEN)["PSNR"]
if b_adv is not None:
    master_psnr["Bilateral"] = b_adv.set_index(SCEN)["PSNR"]

master_psnr = master_psnr.reindex(pd.MultiIndex.from_tuples(list(NOISY.keys()), names=SCEN))
filter_cols = [c for c in master_psnr.columns]
master_psnr["best_method"] = master_psnr[filter_cols].idxmax(axis=1)
master_psnr["best_PSNR"] = master_psnr[filter_cols].max(axis=1)


master_ssim = b_std.pivot_table(index=SCEN, columns="filter", values="SSIM")
if b_nlm is not None:
    master_ssim["NLM"] = b_nlm.set_index(SCEN)["SSIM"]
if b_pm is not None:
    master_ssim["Perona-Malik"] = b_pm.set_index(SCEN)["SSIM"]
if b_adv is not None:
    master_ssim["Bilateral"] = b_adv.set_index(SCEN)["SSIM"]

master_ssim = master_ssim.reindex(pd.MultiIndex.from_tuples(list(NOISY.keys()), names=SCEN))
master_ssim["best_method"] = master_ssim[filter_cols].idxmax(axis=1)
master_ssim["best_SSIM"] = master_ssim[filter_cols].max(axis=1)



params_summary = pd.DataFrame(index=master_psnr.index)
for fname in ["mean", "gaussian", "median"]:
    p_col = b_std[b_std["filter"] == fname].set_index(SCEN)["best_param"]
    params_summary[f"{fname}_param"] = p_col

if b_nlm is not None:
    b_nlm_idx = b_nlm.set_index(SCEN)
    params_summary["nlm_param"] = [f"h={r.h}, p={int(r.patch_size)}" for _, r in b_nlm_idx.iterrows()]
if b_pm is not None:
    b_pm_idx = b_pm.set_index(SCEN)
    params_summary["pm_param"] = [f"opt={int(r.option)}, k={r.kappa}, it={int(r.n_iter)}" for _, r in b_pm_idx.iterrows()]
if b_adv is not None:
    b_adv_idx = b_adv.set_index(SCEN)
    params_summary["bilat_param"] = [f"sc={r.sigma_color}, ss={r.sigma_spatial}" for _, r in b_adv_idx.iterrows()]



master_psnr.round(2).to_csv(TABLE_DIR / "07_master_psnr_summary.csv")
master_ssim.round(3).to_csv(TABLE_DIR / "07_master_ssim_summary.csv")
params_summary.to_csv(TABLE_DIR / "07_best_parameters_summary.csv")
print(f"Master benchmark tables exported to: {TABLE_DIR.resolve()}")

print("\n==================== GLOBAL PSNR SUMMARY (dB) ====================")
display(master_psnr.round(2))

print("\n==================== GLOBAL SSIM SUMMARY ====================")
display(master_ssim.round(3))

print("\n==================== OPTIMAL PARAMETERS PER METHOD ====================")
display(params_summary)

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
images = list(CLEAN.keys())
noises = list(NOISE_CFG.keys())

for r, img_name in enumerate(images):
    for c, n_type in enumerate(noises):
        ax = axes[r, c]
        sub_df = master_psnr.loc[(img_name, n_type)].reindex(LEVEL_ORDER)
        for m in filter_cols:
            ax.plot(sub_df.index, sub_df[m], marker="o", linewidth=2, label=m)
        ax.set_title(f"{img_name} | Noise: {n_type.capitalize()}", fontsize=11, fontweight="bold")
        ax.set_xlabel("Noise Level")
        ax.set_ylabel("PSNR (dB)")
        ax.grid(True, linestyle="--", alpha=0.6)
        if r == 0 and c == 1:
            ax.legend(bbox_to_anchor=(1.05, 1), loc="upper left", title="Filters")

fig.suptitle("Comparative PSNR Evolution Across Noise Levels", fontsize=14, fontweight="bold")
plt.tight_layout()
save_figure(fig, "07_global_psnr_comparison.png")
plt.show()
```
